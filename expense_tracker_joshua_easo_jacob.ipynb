```python

%%writefile app.py
from flask import Flask, render_template, request

app = Flask(__name__)
expenses = []

@app.route("/")
def home():
    return render_template("index.html", expenses=expenses)

@app.route("/about")
def about():
    return "Welcome to the Expense Tracker project!"

@app.route("/health")
def health():
    return "Expense Tracker is running successfully!"

@app.route("/add", methods=["POST"])
def add_expense():
    name = request.form.get("expense_name", "").strip()
    amount_text = request.form.get("amount", "").strip()

    try:
        amount = float(amount_text)
        if not name or amount <= 0:
            return "Invalid input! Enter a name and an amount above zero."
    except ValueError:
        return "Invalid amount! Please enter a valid number."

    expenses.append({"name": name, "amount": amount})
    return render_template("index.html", expenses=expenses)

if __name__ == "__main__":
    app.run(host="0.0.0.0", port=5001,
            debug=False, use_reloader=False)
```

```python

%%writefile templates/index.html
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1">
    <title>My Expense Tracker</title>
</head>
<body>
    <h1>My Expense Tracker</h1>
    <p>Record your daily spending easily.</p>

    <h2>Add an Expense</h2>
    <form action="/add" method="POST">
        <label>Expense Name:</label>
        <input type="text" name="expense_name"
               placeholder="e.g. Lunch" required>
        <br><br>

        <label>Amount (₹):</label>
        <input type="number" name="amount"
               min="0.01" step="0.01" required>
        <br><br>

        <button type="submit">Add Expense</button>
    </form>

    <hr>
    <h2>Recorded Expenses</h2>

    {% if expenses %}
        <ul>
        {% for expense in expenses %}
            <li>{{ expense.name }} — ₹{{ "%.2f"|format(expense.amount) }}</li>
        {% endfor %}
        </ul>

        <h3>Total: ₹{{ "%.2f"|format(expenses|sum(attribute="amount")) }}</h3>
    {% else %}
        <p>No expenses recorded yet.</p>
    {% endif %}

    <p><a href="/about">About this project</a></p>
</body>
</html>
```

```python
%%writefile README.md
# My Expense Tracker

## Project Description
A simple expense tracking web application built using Python and Flask.

## Features
- Homepage to record expenses
- Add expenses using a POST form
- Input validation
- Display recorded expenses
- Calculate total expenses
- About and health routes

## Technologies
- Python
- Flask
- HTML
- Jinja templates

## Installation
Install dependencies:
pip install -r requirements.txt

Run the application:
python app.py

Open http://127.0.0.1:5001 in your browser.

## Routes
- GET / : Homepage
- GET /about : About page
- GET /health : Health check
- POST /add : Add an expense

## Limitation
Expenses are stored temporarily in memory and are cleared when the server restarts.
```

```python

import os
import zipfile

# Create the project folder
os.makedirs("expense_tracker/templates", exist_ok=True)

# Copy the project files into the correct structure
import shutil

shutil.copy("app.py", "expense_tracker/app.py")
shutil.copy("requirements.txt", "expense_tracker/requirements.txt")
shutil.copy("README.md", "expense_tracker/README.md")
shutil.copy(
    "templates/index.html",
    "expense_tracker/templates/index.html"
)

# Package the project
with zipfile.ZipFile(
    "expense_tracker_submission.zip",
    "w",
    zipfile.ZIP_DEFLATED
) as zipf:
    for folder, _, filenames in os.walk("expense_tracker"):
        for filename in filenames:
            filepath = os.path.join(folder, filename)
            zipf.write(filepath, filepath)

print("Project packaged successfully!")
print("Files included:")
for root, _, filenames in os.walk("expense_tracker"):
    for filename in filenames:
        print(os.path.join(root, filename))
```
